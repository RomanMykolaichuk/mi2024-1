# Практичне завдання 4.10
## Практичне використання методів глибокого навчання в індивідуальних (групових) проектах

### Формат
Виконайте **індивідуальний** або **груповий** mini-project. Результатом має бути не лише trained model, а **reviewable evidence package**: problem contract, data provenance, baseline, experiment protocol, metrics, error analysis, reproducible code/config і короткий аналітичний висновок.

## 1. Обов'язкові етапи

1. **Problem statement** — 5–8 речень: input, target, user, success criterion.
2. **Data card** — source/provenance, schema/shape, volume, labels, limitations, data sensitivity.
3. **Split protocol** — train/validation/test або інший обґрунтований evaluation protocol.
4. **Baseline** — просте правило або не-DL / проста модель.
5. **DL model** — обґрунтуйте architecture, loss, optimizer, metric.
6. **Controlled experiments** — не менше 3 runs, змінюючи один ключовий фактор за раз.
7. **Final evaluation** — test використовується після завершення tuning.
8. **Error analysis** — confusion matrix / residuals / типові помилки.
9. **Reproducibility** — seed, config, dependencies, run instructions.
10. **Analytical brief** — що отримано, де модель працює/не працює, limitations, next step.

## 2. Варіанти проектів

Використовуйте лише synthetic/open/дозволені дані. Для чутливих проектів — лише локальне середовище та правила доступу, визначені організацією.

### Варіант 1 — Image classification
Класифікація синтетичних або відкритих зображень. Порівняйте простий baseline з CNN. Перевірте augmentation і domain shift.

### Варіант 2 — Tabular classification з MLP
Побудуйте MLP для табличної задачі й порівняйте з logistic regression / tree baseline. Головне питання: чи дає DL реальну перевагу?

### Варіант 3 — Time-series classification / forecasting
Synthetic/open time series. Використайте temporal split. Порівняйте naive baseline з 1D-CNN/LSTM. Окремо перевірте drift і horizon.

### Варіант 4 — Anomaly detection
Synthetic sensor-like data. Порівняйте простий threshold/statistical baseline з autoencoder або neural classifier. Аналізуйте false alarms.

### Варіант 5 — Text classification
Open/synthetic короткі тексти. Порівняйте bag-of-words baseline із neural/Transformer-based representation. Зафіксуйте provenance та class balance.

### Варіант 6 — Власний проект
Допускається власна задача за умови, що вона безпечна, має явний data source, baseline, evaluation protocol і може бути відтворена викладачем/іншою групою.

## 3. Індивідуальний проект

### Мінімальна структура репозиторію
```text
project/
├── README.md
├── data/README.md           # provenance, не обов'язково самі дані
├── notebooks/
│   ├── 01_data.ipynb
│   ├── 02_baseline.ipynb
│   └── 03_dl_experiment.ipynb
├── src/                     # за потреби
├── config.json              # або YAML
├── requirements.txt
└── results/
    ├── metrics.csv
    └── error_analysis.md
```

У README обов'язково вкажіть: задачу, дані, спосіб запуску, baseline, final result, limitations.

## 4. Груповий проект

### Мінімальні ролі
- **Analytics lead** — problem contract, success criteria, final brief.
- **Data lead** — data card, split, preprocessing, leakage checks.
- **Model lead** — architecture, training pipeline, checkpoints.
- **Evaluation lead** — baseline comparison, metrics, error analysis.
- **Integration/reproducibility lead** — branches/PR, config, environment, run instructions.

За малої групи ролі можна поєднувати. За великої — одну роль можуть виконувати кілька осіб.

### Обов'язкові group evidence
- один спільний project contract;
- issues/tasks або інший явний розподіл робіт;
- branch/commit history;
- мінімум один peer review / pull request review;
- спільна experiment table;
- інтегрований final run;
- 5–7 хвилинний командний review brief.

## 6. Peer-review checklist

Інша особа / група повинна відповісти:

1. Чи зрозуміло, яке аналітичне питання вирішує model?
2. Чи можна відтворити data split?
3. Чи baseline порівнюється на тому самому protocol?
4. Чи test не використовувався для tuning?
5. Чи architecture обґрунтована даними і задачею?
6. Чи є evidence overfit/underfit?
7. Чи показані failure cases?
8. Чи задокументовані seeds, versions, hyperparameters?
9. Чи відсутні credentials/sensitive data у GitHub?
10. Чи висновок відповідає evidence і не перебільшує можливості моделі?

## 7. Критерії оцінювання — 100 балів

| Компонент | Бали |
|---|---:|
| Problem contract і доречність DL | 10 |
| Data provenance, split, leakage control | 15 |
| Baseline | 10 |
| Архітектура і training design | 15 |
| Controlled experiments | 15 |
| Final evaluation + error analysis | 15 |
| Reproducibility / Git evidence | 10 |
| Analytical conclusion + limitations | 10 |

Для групового проекту оцінюється також узгодженість внесків і інтеграція. Високий score моделі без baseline, provenance або error analysis не може компенсувати методичні прогалини.

## 8. Що здати
